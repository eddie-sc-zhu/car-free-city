```python
import pandas as pd
from Model.main import PublicTransitModel

model = PublicTransitModel(data_path="../datasets/daily_ridership.csv",
                 initial_revenue=0,
                 initial_services=1,
                 initial_convenience=1,
                 fare=1,
                 service_coeff=0.1,
                 convenience_effect=5,
                 travel_time_factor=1)

for i in range(len(model.ridership_data)):
    model.step()

results = model.datacollector.get_model_vars_dataframe()
results

zeros_row = pd.DataFrame([[0] * len(model.ridership_data.columns)], columns=model.ridership_data.columns)

# Prepend the zeros row to the original DataFrame
df = pd.concat([zeros_row, model.ridership_data], ignore_index=True)

merged_df = pd.merge(results, df, left_index=True, right_index=True)
#merged_df
```

```python
import matplotlib.pyplot as plt

x = merged_df['Ridership_x']
y = merged_df['Ridership_y']
plt.plot(x)
plt.plot(y)
plt.ylim(0, y.max())
```
